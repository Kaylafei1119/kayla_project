#### Gradient Descent

Every neural network, from a simple linear model to GPT-4, learns by gradient descent. Before we train any models, let's build the core optimization algorithm from scratch.

Minimize $f(x) = x^2$ using gradient descent. The minimum is obviously $x=0$, but you must arrive at it iteratively, the same way a neural network adjusts millions of parameters during training. 

The derivative of $f(x)=x^2$ is $f'(x)=2x$. At each step, update $x$ using the gradient descent rule:
$$x_{new} = x_{old} - \alpha \cdot f'(x_{old})$$

where  $\alpha$ is the learning rate, a small constant that controls step size. Gradient descent following the steepest path downhill.

<img src="grad_descent_viz.png" width="300" height="150">

Input:
* **iterations** - how many gradient descent steps to execute. Can be zero (meaning return the initial value unchanged). iterations >= 0.
* **learning_rate** - the step-size multiplier $\alpha$ that scales each gradient update. Strictly between 0 and 1.
* **init** - the starting point for optimization. Can be any number (including zero).

Output: Apply the update rule for the specified number of iterations starting from init, then return the final value of $x$.

```python
def get_minimizer(self, iterations: int, learning_rate: float, init: int) -> float:
    # Objective function: f(x) = x^2
    # Derivative:         f'(x) = 2x
    # Update rule:        x = x - learning_rate * f'(x)
    currPoint = init
    for i in range(iterations):
        derivative = 2 * currPoint
        currPoint = currPoint- learning_rate * derivative
    # Round final output to 5 decimal places
    optimalPoint = round(currPoint, 5)
    return optimalPoint
```

#### Sigmoid & ReLU

Activation functions are what give neural networks the ability to learn complex patterns. Without them, a neural network is just a series of linear transformations. No matter how many layers you stack, the output would still be a linear function of the input.

Think of activation functions as the decision-makers in a neural network. Each neuron computes a weighted sum of its inputs, but without an activation function, the neuron can only model linear relationships. Activation functions introduce non-linearity, which is what allows neural networks to learn things like image recognition and language understanding.

Implement two of the most common activation functions from scratch using NumPy:
* **Sigmoid**: Squishes any input to a value between 0 and 1. Used for probabilities and binary classification. The formula is: $\sigma(x)=\frac{1}{1+e^{(-x)}}$.
    * Sigmoid was historically popular because it maps values to a nice (0, 1) range, making it easy to interpret as a probability. However, it suffers from the vanishing gradient problem: for very large or very small inputs, the gradient is nearly zero, which slows down learning.
* **ReLU** (Rectified Linear Unit): Returns 0 for negative inputs and the input itself for positive values. The formula is: $\text{ReLu}(x)=\max(0,x)$
    * ReLU largely solved this problem. It's computationally cheap and doesn't saturate for positive values. It's become the default activation function in most modern neural networks.

<img src="sigmoid_curve.png" width="400" height="300">
The sigmoid curve squishes any input to $[0, 1]$.

```python
def sigmoid(self, z: NDArray[np.float64]) -> NDArray[np.float64]:
    # z is a 1D NumPy array
    # Formula: 1 / (1 + e^(-z))
    # return np.round(output, 5)
    sig = 1 / (1 + np.exp(-z))
    return np.round(sig, 5)

def relu(self, z: NDArray[np.float64]) -> NDArray[np.float64]:
    # z is a 1D NumPy array
    # Formula: max(0, z) element-wise
    # np.maximum(arr1, arr2) to compare two arrays element-wise
    relu = np.maximum(0, z)
    return relu
```

#### Softmax

Softmax converts a vector of raw scores (called logits) into a probability distribution. The output values are all positive and sum to 1, making them interpretable as probabilities.

This is how models like GPT decide which token comes next. The final layer outputs a score for every possible token, and softmax turns those scores into probabilities.

The formula for the $i$-th element is: $\text{softmax}(z)_i = \frac{e^{z_i}}{\sum_j}$
